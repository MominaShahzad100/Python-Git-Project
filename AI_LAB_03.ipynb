```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris

# Load Iris Dataset
iris = load_iris()

# Create DataFrame
df = pd.DataFrame(iris.data, columns=iris.feature_names)

# Add target
df["target"] = iris.target

# Add class names
df["class"] = df["target"].map({
    0: "setosa",
    1: "versicolor",
    2: "virginica"
})

print("Dataset loaded successfully!")
print(df.head())
```

```python
# Q1 - Cell 2
# Identify number of records and features

number_of_records = df.shape[0]
number_of_features = len(iris.feature_names)

print("Number of Records:", number_of_records)
print("Number of Features:", number_of_features)

print("\nFeature Names:")
for feature in iris.feature_names:
    print("-", feature)

print("\nTarget Variable: class")
print("Target Classes:", df["class"].unique())
```

```python
# Q1 - Cell 3
# Display data types

print("Data Types:")
print(df.dtypes)

print("\nDataset Information:")
df.info()

print("\nBasic Statistical Information:")
print(df.describe())
```

```python
# Q1 - Cell 4
# NumPy statistical calculations

# Select only numerical feature data
X = df[iris.feature_names].values

# Calculate statistics using NumPy
mean_values = np.mean(X, axis=0)
min_values = np.min(X, axis=0)
max_values = np.max(X, axis=0)
std_values = np.std(X, axis=0)

# Display results
statistics = pd.DataFrame({
    "Feature": iris.feature_names,
    "Mean": mean_values,
    "Minimum": min_values,
    "Maximum": max_values,
    "Standard Deviation": std_values
})

print(statistics)
```

```python
# Q1 - Cell 5
# Select relevant columns

selected_columns = df[
    [
        "sepal length (cm)",
        "sepal width (cm)",
        "petal length (cm)",
        "petal width (cm)",
        "class"
    ]
]

print("Selected Columns:")
print(selected_columns.head(10))
```

```python
# Q1 - Cell 6
# Filter records

# Example: flowers having petal length greater than 4.5 cm
filtered_data = df[df["petal length (cm)"] > 4.5]

print("Flowers with petal length greater than 4.5 cm:")
print(filtered_data.head(10))

print("\nNumber of filtered records:", len(filtered_data))
```

```python
# Q1 - Cell 7
# Group data according to Iris flower class

grouped_data = df.groupby("class")[iris.feature_names].mean()

print("Mean feature values for each Iris class:")
print(grouped_data)
```

```python
# Q1 - Cell 8
# Calculate summary statistics for each class

summary_statistics = df.groupby("class")[iris.feature_names].agg(
    ["mean", "min", "max", "std"]
)

print(summary_statistics)
```

```python
# Q1 - Cell 9
# Visualization 1: Sepal Length vs Sepal Width

plt.figure(figsize=(8, 6))

for class_name in df["class"].unique():
    class_data = df[df["class"] == class_name]

    plt.scatter(
        class_data["sepal length (cm)"],
        class_data["sepal width (cm)"],
        label=class_name
    )

plt.xlabel("Sepal Length (cm)")
plt.ylabel("Sepal Width (cm)")
plt.title("Sepal Length vs Sepal Width")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Q1 - Cell 10
# Visualization 2: Petal Length vs Petal Width

plt.figure(figsize=(8, 6))

for class_name in df["class"].unique():
    class_data = df[df["class"] == class_name]

    plt.scatter(
        class_data["petal length (cm)"],
        class_data["petal width (cm)"],
        label=class_name
    )

plt.xlabel("Petal Length (cm)")
plt.ylabel("Petal Width (cm)")
plt.title("Petal Length vs Petal Width")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Q1 - Cell 11
# Visualization 3: Distribution of Iris Features

plt.figure(figsize=(10, 6))

plt.hist(
    df["sepal length (cm)"],
    bins=15,
    alpha=0.6,
    label="Sepal Length"
)

plt.hist(
    df["petal length (cm)"],
    bins=15,
    alpha=0.6,
    label="Petal Length"
)

plt.xlabel("Measurement (cm)")
plt.ylabel("Frequency")
plt.title("Distribution of Sepal Length and Petal Length")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Q1 - Cell 12
# Visualization 4: Average feature values by class

class_means = df.groupby("class")[iris.feature_names].mean()

class_means.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.xlabel("Iris Class")
plt.ylabel("Average Measurement (cm)")
plt.title("Average Feature Values by Iris Class")
plt.xticks(rotation=0)
plt.legend(title="Features")
plt.grid(axis="y")
plt.show()
```

```python
# Q1 - Cell 13
# Display class-wise means for analysis

print("Class-wise Mean Values:\n")
print(class_means)

print("\nKey Observations:")
print("1. Petal length and petal width show clear differences among Iris classes.")
print("2. Setosa generally has smaller petal measurements than Versicolor and Virginica.")
print("3. Versicolor and Virginica have more similar sepal measurements than their petal measurements.")
print("4. Petal length and petal width appear to be useful features for distinguishing Iris classes.")
```

```python
# Q2 - Cell 1
# Import libraries and load Iris dataset

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# Load dataset
iris = load_iris()

# Features
X = iris.data

# Target
y = iris.target

# DataFrame
df = pd.DataFrame(X, columns=iris.feature_names)

df["target"] = y

df["class"] = df["target"].map({
    0: "setosa",
    1: "versicolor",
    2: "virginica"
})

print("Dataset loaded successfully!")
print(df.head())
```

```python
# Q2 - Cell 2

print("Input Features:")
for feature in iris.feature_names:
    print("-", feature)

print("\nTarget Variable: Iris Flower Class")

print("\nNumber of Input Features:", X.shape[1])
print("Number of Records:", X.shape[0])

print("\nMachine Learning Problem: Supervised Classification")

print("\nExpected Output:")
print("The model predicts whether a flower is:")
print("1. Setosa")
print("2. Versicolor")
print("3. Virginica")
```

```python
# Q2 - Cell 3
# Split data into training and testing sets

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training records:", len(X_train))
print("Testing records:", len(X_test))
```

```python
# Q2 - Cell 4
# K-Nearest Neighbors Classification

model = KNeighborsClassifier(n_neighbors=5)

# Train model
model.fit(X_train, y_train)

# Make predictions
y_pred = model.predict(X_test)

print("Supervised Learning Model trained successfully!")
```

```python
# Q2 - Cell 5

accuracy = accuracy_score(y_test, y_pred)

print("Model Accuracy:", accuracy)

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_pred,
        target_names=iris.target_names
    )
)
```

```python
# Q2 - Cell 6
# Unsupervised Learning preparation

# Use only numerical features
X_unsupervised = df[iris.feature_names]

print("Features used for clustering:")
print(X_unsupervised.head())

print("\nTarget labels are NOT used for clustering.")
```

```python
# Q2 - Cell 7

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X_unsupervised)

print("Features standardized successfully.")
```

```python
# Q2 - Cell 8
# Unsupervised Learning using K-Means

kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

cluster_labels = kmeans.fit_predict(X_scaled)

# Add cluster information to a copy of the dataset
clustered_df = df.copy()

clustered_df["cluster"] = cluster_labels

print("K-Means clustering completed successfully.")

print("\nFirst 10 records with cluster assignments:")
print(
    clustered_df[
        iris.feature_names + ["cluster"]
    ].head(10)
)
```

```python
# Q2 - Cell 9

cluster_means = clustered_df.groupby("cluster")[iris.feature_names].mean()

print("Mean feature values for each cluster:")
print(cluster_means)
```

```python
# Q2 - Cell 10

plt.figure(figsize=(8, 6))

for cluster in sorted(clustered_df["cluster"].unique()):

    cluster_data = clustered_df[
        clustered_df["cluster"] == cluster
    ]

    plt.scatter(
        cluster_data["petal length (cm)"],
        cluster_data["petal width (cm)"],
        label=f"Cluster {cluster}"
    )

plt.xlabel("Petal Length (cm)")
plt.ylabel("Petal Width (cm)")
plt.title("K-Means Clustering of Iris Dataset")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Q2 - Cell 11
# Compare actual classes with discovered clusters

comparison = pd.crosstab(
    clustered_df["class"],
    clustered_df["cluster"]
)

print("Actual Iris Classes vs K-Means Clusters:")
print(comparison)
```
